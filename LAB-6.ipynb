Classify news articles using text features and a machine-learning model.

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix

news = [
    "India won the cricket match with a great performance",
    "The football team won the championship",
    "The player scored two goals in the final",
    "The cricket team is preparing for the next tournament",
    "New smartphone technology was launched today",
    "Artificial intelligence is changing the technology industry",
    "The company developed a new computer processor",
    "Scientists created a powerful machine learning system",
    "The government announced new policies",
    "The election campaign started this week",
    "The minister addressed the parliament",
    "Political leaders discussed the new government policy"
]

labels = (
    ["Sports"] * 4 +
    ["Technology"] * 4 +
    ["Politics"] * 4
)

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(news)

X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.25, random_state=42, stratify=labels
)

model = SVC(kernel="linear")
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("NEWS ARTICLE CLASSIFICATION")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

new_news = [
    "The cricket player scored a century",
    "The company launched a new artificial intelligence product",
    "The government announced a new election policy"
]

for article, prediction in zip(new_news, model.predict(vectorizer.transform(new_news))):
    print("\n", article)
    print("Category:", prediction)
```
