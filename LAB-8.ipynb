# Lab 8: GloVe Word Embeddings

Explore pretrained GloVe word vectors and semantically related words.

**Note:** Run the cells from top to bottom. Install any required libraries if prompted.

## Program


```python
import gensim.downloader as api
from sklearn.metrics.pairwise import cosine_similarity

# Load pre-trained GloVe model
glove = api.load("glove-wiki-gigaword-50")

sentence1 = "I love machine learning"
sentence2 = "I enjoy artificial intelligence"

def sentence_vector(sentence):
    words = sentence.lower().split()
    vectors = [glove[word] for word in words if word in glove]
    return sum(vectors) / len(vectors)

vector1 = sentence_vector(sentence1)
vector2 = sentence_vector(sentence2)

similarity = cosine_similarity([vector1], [vector2])[0][0]

print("Sentence 1 :", sentence1)
print("Sentence 2 :", sentence2)
print("Cosine Similarity :", round(similarity, 2))
```
