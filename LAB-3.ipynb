# Lab 3: One-Hot Encoding, Bag of Words and Word Cloud

Represent text using simple vectorization methods and visualize frequent words.

**Note:** Run the cells from top to bottom. Install any required libraries if prompted.

## Program


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
from wordcloud import WordCloud

corpus = [
    "Machine learning is useful",
    "Machine learning is powerful",
    "Python is useful for machine learning"
]

words = sorted(set(" ".join(corpus).lower().split()))
one_hot = []

for i, word in enumerate(words):
    row = [1 if j == i else 0 for j in range(len(words))]
    one_hot.append(row)

df = pd.DataFrame(one_hot, index=words, columns=words)

print("ACTIVITY 1: ONE-HOT ENCODING")
print(df)

vectorizer = CountVectorizer()
bow = vectorizer.fit_transform(corpus)

print("\nACTIVITY 2: BAG OF WORDS")
print("Vocabulary:", vectorizer.get_feature_names_out())
print("\nBoW Matrix:")
print(pd.DataFrame(bow.toarray(), columns=vectorizer.get_feature_names_out()))

text = " ".join(corpus)
wordcloud = WordCloud(width=800, height=400, background_color="white").generate(text)

plt.imshow(wordcloud)
plt.axis("off")
plt.title("Word Cloud")
plt.show()

frequency = bow.toarray().sum(axis=0)
words = vectorizer.get_feature_names_out()
top5 = sorted(zip(words, frequency), key=lambda x: x[1], reverse=True)[:5]

print("\nTop 5 Most Frequent Words:")
for word, count in top5:
    print(word, ":", count)
```
