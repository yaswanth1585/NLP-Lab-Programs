```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix

# Activity 1: Spam Detection
emails = [
    "Win a free prize now",
    "Congratulations you won a lottery",
    "Claim your free reward",
    "You have won a cash prize",
    "Get free money today",
    "Meeting scheduled for tomorrow",
    "Please submit the project report",
    "Your class starts at 10 AM",
    "Can we meet today",
    "Please call me after class"
]
labels = ["spam"] * 5 + ["ham"] * 5

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(emails)
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.3, random_state=42, stratify=labels
)

model = MultinomialNB()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("ACTIVITY 1: SPAM DETECTION")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

new_emails = ["You won a free prize", "Please attend the meeting tomorrow"]
for email, result in zip(new_emails, model.predict(vectorizer.transform(new_emails))):
    print(email, "->", result)

# Activity 2: Sentiment Analysis
reviews = [
    "The product is excellent",
    "I really loved this product",
    "Amazing quality and service",
    "Very good product",
    "I am happy with the purchase",
    "The product is terrible",
    "I hate this product",
    "Very poor quality",
    "Worst product ever",
    "I am disappointed with the purchase"
]
sentiments = ["positive"] * 5 + ["negative"] * 5

vectorizer2 = TfidfVectorizer()
X2 = vectorizer2.fit_transform(reviews)
X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, sentiments, test_size=0.3, random_state=42, stratify=sentiments
)

svm = SVC(kernel="linear")
svm.fit(X2_train, y2_train)
y2_pred = svm.predict(X2_test)

print("\nACTIVITY 2: SENTIMENT ANALYSIS")
print("Accuracy:", accuracy_score(y2_test, y2_pred))
print("Confusion Matrix:\n", confusion_matrix(y2_test, y2_pred))

new_reviews = ["This product is very good", "The product quality is very bad"]
for review, result in zip(new_reviews, svm.predict(vectorizer2.transform(new_reviews))):
    print(review, "->", result)
```
