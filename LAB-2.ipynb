## In-Lab Program


```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag, ne_chunk

for package in [
    'punkt', 'punkt_tab', 'stopwords', 'wordnet',
    'averaged_perceptron_tagger', 'averaged_perceptron_tagger_eng',
    'maxent_ne_chunker', 'maxent_ne_chunker_tab', 'words'
]:
    nltk.download(package, quiet=True)

sentence = "Dr. Ravi is studying Machine Learning at Anna University in Chennai."

tokens = word_tokenize(sentence)
print("1. Tokenization:")
print(tokens)

stop_words = set(stopwords.words('english'))
cleaned = [w for w in tokens if w.lower() not in stop_words
           and w not in string.punctuation]

print("\n2. Stop-word & Punctuation Removal:")
print(cleaned)

stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in cleaned]
print("\n3. Stemming:")
print(stemmed)

lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in cleaned]
print("\n4. Lemmatization:")
print(lemmatized)

pos = pos_tag(lemmatized)
print("\n5. POS Tagging:")
print(pos)

print("\n6. Named Entity Recognition:")
print(ne_chunk(pos))
```

## Post-Lab Program


```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('wordnet', quiet=True)

sentences = [
    "The students are studying different subjects.",
    "The children are playing in the gardens.",
    "He studies machines and technologies.",
    "The runners were running quickly.",
    "The companies are developing better products."
]

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

for sentence in sentences:
    words = word_tokenize(sentence)
    print("Sentence:", sentence)
    print("Stemming:", [stemmer.stem(w) for w in words])
    print("Lemmatization:", [lemmatizer.lemmatize(w) for w in words])
    print()

words = [
    "studying", "students", "children", "playing",
    "gardens", "studies", "machines", "technologies",
    "running", "runners", "companies", "developing",
    "products", "better"
]

print("STEMMING vs LEMMATIZATION")
for word in words:
    print(word, "->", stemmer.stem(word), "|", lemmatizer.lemmatize(word))
```
