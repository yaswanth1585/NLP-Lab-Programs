# Lab 4: TF-IDF

Convert documents into TF-IDF feature vectors.

**Note:** Run the cells from top to bottom. Install any required libraries if prompted.

## In-Lab Program


```python
import pandas as pd
import math

corpus = [
    "machine learning is useful",
    "machine learning is powerful",
    "python is useful for machine learning"
]

vocabulary = sorted(set(" ".join(corpus).split()))
print("ACTIVITY 1: VOCABULARY")
print(vocabulary)

print("\nTotal words in each document:")
for i, doc in enumerate(corpus):
    print("Document", i + 1, ":", len(doc.split()))

tf = []
for doc in corpus:
    words = doc.split()
    total = len(words)
    tf.append([words.count(word) / total for word in vocabulary])

tf_df = pd.DataFrame(tf, columns=vocabulary)
print("\nTF MATRIX:")
print(tf_df.round(3))

df = [sum(word in doc.split() for doc in corpus) for word in vocabulary]
N = len(corpus)
idf = [math.log(N / value) for value in df]

df_idf = pd.DataFrame({"Word": vocabulary, "DF": df, "IDF": idf})
print("\nDF AND IDF:")
print(df_idf.round(3))

tfidf = [[tf[i][j] * idf[j] for j in range(len(vocabulary))]
         for i in range(len(corpus))]
print("\nTF-IDF MATRIX:")
print(pd.DataFrame(tfidf, columns=vocabulary).round(3))
```

## Post-Lab Program


```python
import pandas as pd
import math

corpus = [
    "Python is easy",
    "Python supports machine learning",
    "Machine learning is powerful"
]

vocabulary = sorted(set(" ".join(corpus).lower().split()))
print("Vocabulary:", vocabulary)

tf = []
for doc in corpus:
    words = doc.lower().split()
    total = len(words)
    tf.append([words.count(word) / total for word in vocabulary])

print("\nTF Matrix:")
print(pd.DataFrame(tf, columns=vocabulary).round(3))

df = [sum(word in doc.lower().split() for doc in corpus) for word in vocabulary]
N = len(corpus)
idf = [math.log(N / x) for x in df]

print("\nDF and IDF:")
print(pd.DataFrame({"Word": vocabulary, "DF": df, "IDF": idf}).round(3))

tfidf = [[tf[i][j] * idf[j] for j in range(len(vocabulary))]
         for i in range(len(corpus))]

print("\nTF-IDF Matrix:")
print(pd.DataFrame(tfidf, columns=vocabulary).round(3))
```
